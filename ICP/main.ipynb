```python
import open3d as o3d
import numpy as np

# Load point clouds
pcd1 = o3d.io.read_point_cloud("/home/ashutosh/Documents/ICP/icp_example/build/aligned.pcd")
pcd2 = o3d.io.read_point_cloud("/home/ashutosh/Documents/ICP/icp_example/build/output1.pcd")
pcd3 = o3d.io.read_point_cloud("/home/ashutosh/Documents/ICP/icp_example/build/output2.pcd")

# Assign unique colors (R,G,B in [0,1])
pcd1.paint_uniform_color([1, 0, 0])  # Red
pcd2.paint_uniform_color([0, 1, 0])  # Green
pcd3.paint_uniform_color([0, 0, 1])  # Blue

# Visualize
o3d.visualization.draw_geometries([pcd1, pcd2, pcd3])
```
